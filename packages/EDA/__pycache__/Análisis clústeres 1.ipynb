```python
%matplotlib inline
from sklearn.datasets import make_blobs
from yellowbrick.cluster import KElbowVisualizer
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
import numpy as np
from sklearn.cluster import KMeans, AgglomerativeClustering

import seaborn as sns
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('df_limpio_con_clusters.csv')
df['cluster_kmeans_3'].value_counts()
df['cluster_kmeans_3'].value_counts(normalize=True)
num_cols = df.select_dtypes(include='number').columns
cat_cols = df.select_dtypes(include=['object', 'category']).columns
```

```python
cluster_profile_num = (
    df
    .groupby('cluster_kmeans_3')[num_cols]
    .mean()
)
cluster_profile_num

cluster_profile_median = (
    df
    .groupby('cluster_kmeans_3')[num_cols]
    .median()
)
print("PERFIL MEDIO POR CLÚSTER")
print(cluster_profile_num.to_string())

print("\nPERFIL MEDIANO POR CLÚSTER")
print(cluster_profile_median.to_string())
```

```python
df_clusters = cluster_profile_num.copy()
paleta =["#F48A48","#6F4E37","#EBB68D","#391704", "#FF4704"]
```

```python
def barplot_cluster(df, y, title, ylabel):
    plt.figure(figsize=(7,5))
    sns.barplot(
        data=df,
        x='cluster_kmeans_3',
        y=y,
        palette=paleta,
        errorbar=None
    )
    plt.title(title)
    plt.xlabel('Clúster')
    plt.ylabel(ylabel)
    plt.show()

#Sin más
barplot_cluster(
    df_clusters,
    'lead_time',
    'Lead time medio por clúster',
    'Días'
)

barplot_cluster(
    df_clusters,
    'lenght_of_stay',
    'Duración media de la estancia por clúster',
    'Noches'
)


barplot_cluster(
    df_clusters,
    'total_personas_adult_y_niños',
    'Número medio de personas por clúster',
    'Personas'
)

#Más o menos bueno
barplot_cluster(
    df_clusters,
    'reservation_net_value',
    'Valor neto medio de la reserva por clúster',
    '€'
)

#Más o menos bueno
barplot_cluster(
    df_clusters,
    'total_adr',
    'ADR medio por clúster',
    '€'
)

#Más o menos bueno
barplot_cluster(
    df_clusters,
    'cancellation_lead_time',
    'Anticipación media de cancelación por clúster',
    'Días antes del check-in'
)
```

```python
#BUENO
fig, ax1 = plt.subplots(figsize=(8,5))

sns.barplot(
    data=df_clusters,
    x='cluster_kmeans_3',
    y='reservation_net_value',
    palette=paleta,
    ax=ax1,
    errorbar=None
)

ax1.set_ylabel('Valor neto medio (€)')
ax1.set_xlabel('Clúster')

ax2 = ax1.twinx()
sns.pointplot(
    data=df_clusters,
    x='cluster_kmeans_3',
    y='cancellation_lead_time',
    color='black',
    ax=ax2
)

ax2.set_ylabel('Días antes del check-in')
plt.title('Valor económico vs anticipación de cancelación por clúster')
plt.show()
```

```python
#BUENO
import numpy as np

vars_radar = [
    'lead_time',
    'lenght_of_stay',
    'total_personas_adult_y_niños',
    'reservation_net_value',
    'total_adr',
    'cancellation_lead_time'
]

df_radar = df_clusters[['cluster_kmeans_3'] + vars_radar]
df_norm = df_radar.copy()

# Normalización 0-1
for v in vars_radar:
    df_norm[v] = (df_norm[v] - df_norm[v].min()) / (df_norm[v].max() - df_norm[v].min())

labels = vars_radar
angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(7,7))
ax = plt.subplot(111, polar=True)

for c in df_norm['cluster_kmeans_3']:
    values = df_norm[df_norm['cluster_kmeans_3'] == c][labels].values.flatten().tolist()
    values += values[:1]
    ax.plot(angles, values, label=f'Clúster {c}')
    ax.fill(angles, values, alpha=0.1)

ax.set_thetagrids(np.degrees(angles[:-1]), labels)
plt.title('Perfil normalizado de los clústeres')
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.show()
```

```python
def cat_distribution(df, cat_col):
    return (
        pd.crosstab(df['cluster_kmeans_3'], df[cat_col], normalize='index') * 100
    )

cat_distribution(df, 'status')
cat_distribution(df, 'rate_type')
cat_distribution(df, 'business_segment')
```
