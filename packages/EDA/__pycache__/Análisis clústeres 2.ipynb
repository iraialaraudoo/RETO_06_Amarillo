```python
%matplotlib inline
from sklearn.datasets import make_blobs
from yellowbrick.cluster import KElbowVisualizer
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
import numpy as np
from sklearn.cluster import KMeans, AgglomerativeClustering

import seaborn as sns
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('df_limpio_con_clusters.csv')
```

```python
paleta =["#F48A48","#6F4E37","#EBB68D","#391704", "#FF4704"]
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x='lead_time',
    hue='cluster_kmeans_3',
    bins=30,
    multiple='layer',
    palette=paleta,
    alpha=0.6
)

plt.title('Distribución del lead time por clúster')
plt.xlabel('Lead time (días)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x='reservation_net_value',
    hue='cluster_kmeans_3',
    bins=30,
    multiple='layer',
    palette=paleta,
    alpha=0.6
)

plt.title('Distribución del valor neto de la reserva por clúster')
plt.xlabel('Valor neto (€)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x='total_adr',
    hue='cluster_kmeans_3',
    bins=30,
    multiple='layer',
    palette=paleta,
    alpha=0.6
)

plt.title('Distribución del ADR por clúster')
plt.xlabel('ADR (€)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x='total_personas_adult_y_niños',
    hue='cluster_kmeans_3',
    bins=15,
    multiple='stack',
    palette=paleta,
    alpha=0.8
)

plt.title('Distribución del número de personas por clúster')
plt.xlabel('Número de personas')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    data=df,
    x='cluster_kmeans_3',
    y='reservation_net_value',
    palette=paleta
)

plt.title('Valor neto de la reserva por clúster')
plt.xlabel('Clúster')
plt.ylabel('Valor neto (€)')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    data=df,
    x='cluster_kmeans_3',
    y='total_adr',
    palette=paleta
)

plt.title('ADR por clúster')
plt.xlabel('Clúster')
plt.ylabel('ADR (€)')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.barplot(
    data=df,
    x='cluster_kmeans_3',
    y='total_personas_adult_y_niños',
    palette=paleta,
    errorbar=None
)

plt.title('Número medio de personas por clúster')
plt.xlabel('Clúster')
plt.ylabel('Personas por reserva')
plt.show()
```
